```python
corpus = [
    "The cat sat on the mat.",
    "The dog sat on the log.",
    "Cats and dogs are common pets."
]
```

### Bag of Words (BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(corpus)

print("BoW Vocabulary:", vectorizer.get_feature_names_out())
print("BoW Vectors:\n", bow_matrix.toarray())
```

### TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(corpus)

print("TF-IDF Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Vectors:\n", tfidf_matrix.toarray())
```

### Word2Vec

```python
import fasttext
import os


with open("temp_fasttext.txt", "w") as f:
    f.write("\n".join(corpus))

model = fasttext.train_unsupervised("temp_fasttext.txt", model='skipgram', dim=50, minCount = 1)

print("Words: ", model.words)
for word in model.words:
    vec = model.get_word_vector(word)
    print(f"Word: {word} → Vector Dim: {len(vec)} -> Vector (first 5 dims): {vec[:5]}")

os.remove("temp_fasttext.txt")
```

### MiniLM

```python
# Install first (if not already done): pip install -U sentence-transformers
from sentence_transformers import SentenceTransformer

# Initialize BERT-based sentence embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Get embeddings (1 vector per sentence)
embeddings = model.encode(corpus)

print("\nSentence Embedding for:", corpus[0])
print(f"embedding: DIm f{len(embeddings[0])} -> vector : first 5 {embeddings[0][:5]}" )

print("\nSentence Embedding for:", corpus[1])
print(f"embedding: DIm f{len(embeddings[1])} -> vector : first 5 {embeddings[1][:5]}" )

```
